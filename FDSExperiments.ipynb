```python
# EXPERIMENT 4
# Count the frequency of occurrence of a word in a body of text

import re
from collections import Counter

# Sample text
text = """
Data science is a growing field.
Data science uses Python for data analysis.
Python is popular in data science.
Data science includes machine learning and data visualization.
Python is widely used for data science.
"""

# Convert text to lowercase
text = text.lower()

# Extract words from the text
words = re.findall(r'\b\w+\b', text)

# Count frequency of each word
word_frequency = Counter(words)

# Display all word frequencies
print("Word Frequencies:")
for word, count in word_frequency.items():
    print(word, ":", count)

# Ask the user for a word
search_word = input("\nEnter a word to find its frequency: ").lower()

# Display frequency of the selected word
frequency = word_frequency[search_word]

print("\nFrequency of", search_word, "=", frequency)
```

```python
# EXPERIMENT 6
# Time Series Analysis of Website Traffic

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --------------------------------------------------
# 1. Create sample website traffic data
# --------------------------------------------------

data = {
    'Date': [
        '2025-01-01', '2025-02-01', '2025-03-01',
        '2025-04-01', '2025-05-01', '2025-06-01',
        '2025-07-01', '2025-08-01', '2025-09-01',
        '2025-10-01', '2025-11-01', '2025-12-01'
    ],

    'Visitors': [
        12000, 13500, 15000, 14200,
        16500, 18000, 19500, 21000,
        20500, 23000, 25000, 27000
    ]
}

df = pd.DataFrame(data)

# --------------------------------------------------
# 2. Convert Date column to datetime
# --------------------------------------------------

df['Date'] = pd.to_datetime(df['Date'])

print("Website Traffic Dataset:")
print(df)


# --------------------------------------------------
# 3. Set Date as index
# --------------------------------------------------

df.set_index('Date', inplace=True)

print("\nDataset after setting Date as index:")
print(df)


# --------------------------------------------------
# 4. Display basic information
# --------------------------------------------------

print("\nDataset Information:")
print(df.info())


# --------------------------------------------------
# 5. Statistical summary
# --------------------------------------------------

print("\nStatistical Summary:")
print(df.describe())


# --------------------------------------------------
# 6. Find total visitors
# --------------------------------------------------

total_visitors = df['Visitors'].sum()

print("\nTotal Visitors:", total_visitors)


# --------------------------------------------------
# 7. Find average visitors
# --------------------------------------------------

average_visitors = df['Visitors'].mean()

print("Average Monthly Visitors:", round(average_visitors, 2))


# --------------------------------------------------
# 8. Find month with highest traffic
# --------------------------------------------------

highest_traffic = df['Visitors'].idxmax()
highest_value = df['Visitors'].max()

print("\nHighest Traffic:")
print("Month:", highest_traffic.strftime('%B %Y'))
print("Visitors:", highest_value)


# --------------------------------------------------
# 9. Find month with lowest traffic
# --------------------------------------------------

lowest_traffic = df['Visitors'].idxmin()
lowest_value = df['Visitors'].min()

print("\nLowest Traffic:")
print("Month:", lowest_traffic.strftime('%B %Y'))
print("Visitors:", lowest_value)


# --------------------------------------------------
# 10. Calculate monthly change
# --------------------------------------------------

df['Monthly_Change'] = df['Visitors'].diff()

print("\nMonthly Change:")
print(df)


# --------------------------------------------------
# 11. Calculate 3-month moving average
# --------------------------------------------------

df['Moving_Average'] = df['Visitors'].rolling(window=3).mean()

print("\n3-Month Moving Average:")
print(df)


# --------------------------------------------------
# 12. Plot website traffic
# --------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    df.index,
    df['Visitors'],
    marker='o',
    color='blue',
    linewidth=2,
    label='Visitors'
)

plt.title('Website Traffic Over Time')
plt.xlabel('Month')
plt.ylabel('Number of Visitors')

plt.grid(True)
plt.legend()

plt.xticks(rotation=45)

plt.show()


# --------------------------------------------------
# 13. Plot traffic and moving average
# --------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    df.index,
    df['Visitors'],
    marker='o',
    color='blue',
    label='Actual Visitors'
)

plt.plot(
    df.index,
    df['Moving_Average'],
    color='red',
    linewidth=3,
    label='3-Month Moving Average'
)

plt.title('Website Traffic and Moving Average')
plt.xlabel('Month')
plt.ylabel('Number of Visitors')

plt.grid(True)
plt.legend()

plt.xticks(rotation=45)

plt.show()


# --------------------------------------------------
# 14. Bar chart of website traffic
# --------------------------------------------------

plt.figure(figsize=(12, 6))

plt.bar(
    df.index.strftime('%b'),
    df['Visitors'],
    color='green'
)

plt.title('Monthly Website Traffic')
plt.xlabel('Month')
plt.ylabel('Number of Visitors')

plt.show()
```

```python
# Experiment 7: Random Sampling and Sampling Distribution

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create a random population of 1000 students' marks
np.random.seed(42)

population = np.random.normal(
    loc=70,      # Mean
    scale=10,    # Standard deviation
    size=1000
)

# Store population in DataFrame
df = pd.DataFrame({"Marks": population})

print("Population:")
print(df.head())
print("\nPopulation Mean:", df["Marks"].mean())

# Take one random sample of 50 students
sample = df.sample(n=50)

print("\nRandom Sample:")
print(sample.head())
print("\nSample Mean:", sample["Marks"].mean())

# Create sampling distribution
sample_means = []

for i in range(100):
    sample = df.sample(n=50)
    sample_mean = sample["Marks"].mean()
    sample_means.append(sample_mean)

# Store sample means
sampling_distribution = pd.DataFrame({
    "Sample Mean": sample_means
})

print("\nSampling Distribution:")
print(sampling_distribution.head())

print("\nMean of Sample Means:",
      sampling_distribution["Sample Mean"].mean())

# Plot sampling distribution
plt.hist(sample_means, bins=10, edgecolor="black")

plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Sampling Distribution of Sample Means")

plt.show()
```

```python
# Experiment 8: Z Test for the Data using Python

import numpy as np
from scipy.stats import norm

# Sample data
data = np.array([
    72, 68, 75, 70, 69,
    74, 71, 73, 67, 76,
    70, 72, 74, 69, 71
])

# Given population parameters
population_mean = 70
population_std = 5

# Calculate sample size and sample mean
n = len(data)
sample_mean = np.mean(data)

# Calculate Z statistic
z = (sample_mean - population_mean) / (population_std / np.sqrt(n))

# Calculate two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z)))

# Significance level
alpha = 0.05

# Display results
print("Sample Data:")
print(data)

print("\nSample Size:", n)
print("Sample Mean:", sample_mean)
print("Population Mean:", population_mean)
print("Population Standard Deviation:", population_std)

print("\nZ Statistic:", z)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("\nReject the Null Hypothesis (H0)")
    print("There is a significant difference between the sample mean and population mean.")
else:
    print("\nFail to Reject the Null Hypothesis (H0)")
    print("There is no significant difference between the sample mean and population mean.")
```

```python
# Experiment 9: Matplotlib Library Use Cases in Data Science

import matplotlib.pyplot as plt
import numpy as np

# Sample data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 140, 200, 220]

students = [65, 70, 72, 68, 75, 80, 85, 78, 90, 88]

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 8, 10])

# 1. Line Plot - Shows trends over time
plt.figure(figsize=(6, 4))
plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.grid()
plt.show()

# 2. Bar Chart - Compares different categories
plt.figure(figsize=(6, 4))
plt.bar(months, sales)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Comparison")
plt.show()

# 3. Scatter Plot - Shows relationship between variables
plt.figure(figsize=(6, 4))
plt.scatter(x, y)
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Relationship Between X and Y")
plt.show()

# 4. Histogram - Shows distribution of data
plt.figure(figsize=(6, 4))
plt.hist(students, bins=5, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()

# 5. Pie Chart - Shows proportions
subjects = ["Python", "Java", "SQL", "C"]
students_count = [30, 25, 20, 25]

plt.figure(figsize=(6, 4))
plt.pie(students_count, labels=subjects, autopct="%1.1f%%")
plt.title("Student Preference by Subject")
plt.show()
```
